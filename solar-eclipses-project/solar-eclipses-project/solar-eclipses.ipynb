# Solar Eclipses

What is the average duration of total darkness during a total solar eclipse? When did the longest solar eclipse occur?

The file `solar-eclipses.csv` provides information on all solar eclipses from 1901 to 2100, leveraging the ability of scientists to accurately predict the timing and location of future eclipses.

A solar eclipse happens when the Moon moves between the Earth and the Sun, blocking the Sun's light either partially or completely. There are three primary types of solar eclipses:

1. Total Solar Eclipse: The Moon fully covers the Sun, casting a shadow on Earth and revealing the Sun's corona.

2. Partial Solar Eclipse: The Moon obscures only part of the Sun, creating a crescent-shaped appearance.

3. Annular Solar Eclipse: The Moon covers the center of the Sun, leaving a ring-like appearance, known as a "ring of fire," around the edges. This occurs when the Moon is too far from Earth to completely cover the Sun.

The `duration` variable indicates the length of time that the entire moon covers the sun.

**QUESTION 1. When did the longest solar eclipse occur? The longest total eclipse?
	- Hint: convert duration to seconds. You can use `str.replace('s', '')` to remove the 's' from the duration values.**

```python
df['duration'].info
```

The datatype is str (string). This means we have to convert the data to number.

```python
#making a copy of df (split_duration) with new columns (minutes and seconds)
split_duration = df.copy()

split_duration[['minutes', 'seconds']] = split_duration['duration'].str.split('m', expand=True) 
split_duration
```

```python
#replacing 's' with '', NaN with 0
split_duration['seconds'] = split_duration['seconds'].str.replace('s', '')
split_duration[['minutes', 'seconds']] = split_duration[['minutes', 'seconds']].fillna(0)
split_duration
```

```python
#converting minutes and seconds to numeric values & caculating total duration in seconds
split_duration[['minutes', 'seconds']] = split_duration[['minutes', 'seconds']].astype('int')
split_duration['total_duration'] = split_duration['minutes'] * 60 + split_duration['seconds']
split_duration
```

```python
#When did the longest solar eclipse occur?
split_duration.sort_values(by='total_duration', ascending=False).head(1)
```

**ANSWER:** The longest Solar Eclipse happened in December 14, 1955, with a total duration of 12 minutes and 9 seconds

```python
#When did the longest total solar eclipse occur?
split_duration.query('eclipse_type == "Total"').sort_values(by='total_duration', ascending=False).head(1)
```

**ANSWER:** The longest Total Solar Eclipse happened in June 20, 1955, that lasts for 7 minutes and 8 seconds.

**QUESTION 2. What is the average duration of total solar eclipses?**

```python
split_duration.query('eclipse_type == "Total"')['total_duration'].mean()
```

**ANSWER:** the average duration of total duration of total eclipses is approximately 211.96 seconds, or  3 minutes and 31 seconds.
