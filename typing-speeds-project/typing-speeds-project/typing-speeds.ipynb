# PROJECT IDEAS

**1. Remove unnecessary columns, such as PARTICIPANT_ID, to streamline the dataset.**

```python
#creating a copy of the original df (typing)
typing = df.copy()

#removing PARTICIPANT_ID:
typing = typing.drop('PARTICIPANT_ID', axis=1) #axis=1 means 'column', 0 is for rows
```

**2. Rename columns (e.g `AVG_WPM_15` to `wpm`, `ROR` to `ror`, `HAS_TAKEN_TYPING_COURSE` to `course`) for brevity and clarity during analysis.**

```python
name_map = {
    'AGE' : 'age',
    'HAS_TAKEN_TYPING_COURSE' : 'course',
    'COUNTRY' : 'country',
    'LAYOUT' : 'layout',
    'NATIVE_LANGUAGE' : 'language',
    'FINGERS' : 'fingers',
    'KEYBOARD_TYPE' : 'keyboard',
    'ERROR_RATE' : 'error_rate',
    'AVG_WPM_15' : 'wpm',
    'ROR' : 'ror'
}

typing = typing.rename(columns = name_map)
typing
```

# FINGER COUNT ANALYSIS

**Finger Count Analysis 1: Compare typing speeds across groups using different numbers of fingers, excluding the "10+" category for simplicity.**

```python
#dropping rows with 10+ fingers
no_11 = typing.query('fingers != "10+"')

#comparing the average typing speeds across groups using different numbers of fingers.
no_11.groupby('fingers')['wpm'].mean()
```

**Inference:** The result implies that people who use 9-10 fingers types the fastest in general. **But**, this method only groups the participants by number of fingers they use, disregarding the other factors. Meaning, people in a specific group may have different age, keyboard layouts, keyboard types, language, and may or may not have taken typing course. Let's address this... 

**Finger Count Analysis 2: Control for consistency by first filtering to similar `AGE`, `KEYBOARD_LAYOUT`, `NATIVE_LANGUAGE`, `KEYBOARD_TYPE`, and `HAS_TAKEN_TYPING_COURSE` values.**

**For this analysis, let's use the group with the ff criteria:**
- age: 25-59 (adults)
- keyboard_layout: qwerty
- native_language: en
- keyboard_type: laptop
- course: 0 (no)

```python
typing_grp = no_11.query(
    'age >=25 and age <= 59 and '
    'layout == "qwerty" and '
    'language == "en" and '
    'keyboard == "laptop" and '
    'course == 0'
)

typing_grp
```

The original data with 168,594 total number of rows is now filtered into 17,273 rows, representing a specific group of participants: **Adult english speaking participants that use laptop with qwerty keyboard and has not taken any typing course.**

**Finger Count Analysis 3: Exclude participants with high error rates (ERROR_RATE > 3%) to focus on reliable data.**

```python
typing_grp = typing_grp.query('error_rate < 3')
typing_grp
```

**Finger Count Analysis 4: Drop columns after filtering if they now only have a single value.**

Since the data now shows single value for course, layout, language, ad keyboard, we can just remove these columns to simplify the table and  focus more on the analysis.

```python
typing_grp = typing_grp.drop(columns = ['course', 'layout', 'language', 'keyboard'])
typing_grp
```

Running another average comparison across groups of participant that use different number of fingers in typing.

```python
typing_grp.groupby('fingers')['wpm'].mean()
```

**ANSWER:** The results show a similar comparison with the unfiltered data -- People that use 9-10 fingers when typing tend to be faster than those who use fewer fingers. (This takes the other factors into consideration, making the inference and interpretation more specific.)

# ROLLOVER RATIO ANALYSIS

**The Rollover Ratio (`ROR`) represents the proportion of keypresses where a new key is pressed before releasing the previous one.**

**Rollover Ratio Analysis: Compare typing speeds between participants with `ROR` ≤ 20% and those with `ROR` > 80%, keeping `AGE`, `KEYBOARD_TYPE`, `FINGERS`, and other variables constant.**

**For this analysis, let's use the group with the ff criteria:**
- age: 25-59 (adults)
- keyboard: laptop
- fingers: 9-10
- course: 0 (no)
- layout: qwerty
- language: en
- error_rate: < 3.00


```python
# filtering the dataset:
rra = typing.query(
    'age >=25 and age <= 59 and '
    'keyboard == "laptop" and '
    'fingers == "9-10" and ' 
    'course == 0 and '
    'layout == "qwerty" and '
    'language == "en" and '
    'error_rate < 3'
)
rra
```

```python
#grouping the filtered dataset and performing rra
#we can't use groupby, so let's use query
uppergrp_mean = rra.query('ror > 0.8')['wpm'].mean()
uppergrp_mean
```

```python
lowergrp_mean = rra.query('ror <= 0.2')['wpm'].mean()
lowergrp_mean
```

**ANSWER:** The results indicate that adult participants in a specific group (those who are english native speakers that use a laptop with qwerty keyboard, use 9-10 fingers in typing, with less than 3% error rate, and has not taken any typing course) differs in typing speeds when grouped according to ROR (Rollover Ratio). People who has a ROR value of greater than  80% has a group wpm mean of 104.76. Whereas, people with that of less than or equal to 20% has a wpm average of 41.68. **This means that people with high Rollover Ratio (ROR), or high proportion of new keypress before releasing the previous one, has a higher word per minute (WPM) in average compared to those with low ROR.**

# INFLUENCE OF TYPING COURSE

**Influence of Typing Course: Compare typing speeds between participants with a typing course (`HAS_TAKEN_TYPING_COURSE` = 1) and without (`HAS_TAKEN_TYPING_COURSE` = 0), holding other variables such as `KEYBOARD_TYPE`, `AGE` range, and `FINGER_COUNT` constant.**

**For this analysis, let's use the group with the ff criteria:**
- age: 25-59 (adults)
- keyboard: laptop
- fingers: 9-10
- layout: qwerty
- language: en
- error_rate: < 3.00

```python
itc = typing.query(
    'age >=25 and age <= 59 and '
    'keyboard == "laptop" and '
    'fingers == "9-10" and '
    'layout == "qwerty" and '
    'language == "en" and '
    'error_rate < 3'
)
itc
```

```python
#grouping the filtered dataset and comparing the wpm
itc_mean = itc.groupby('course')['wpm'].mean()
itc_mean
```

**ANSWER:** The average word per minute (WPM) of the groups of people who has taken a typing course and those who have not states that people with prior typing training types faster, with an average of 62.91 wpm, than people without, which has an wpm average of 58.95.
