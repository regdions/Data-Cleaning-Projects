# Emoji Sentiment

Are popular emojis generally associated with positive or negative sentiments?

The file `"emoji-sentiment.csv"` provides data on the sentiment associated with various emojis.

Researchers examined 1.6 million tweets across 13 European languages. Each tweet was labeled by annotators as positive (+1), negative (-1), or neutral (0). About 4% of these tweets included emojis.

Columns include:
- `Occurrences [5...max]`: Number of times the emoji appears in the dataset.
- `Position [0...1]`: Average position of the emoji in tweets, from start (0) to end (1).
- `Neg [0...1]`: Percentage of tweets with the emoji that are 'negative'.
- `Neu [0...1]`: Percentage of tweets with the emoji that are 'neutral'.
- `Pos [0...1]`: Percentage of tweets with the emoji that are 'positive'.

**DC1. Remove unnecessary columns that are not useful for your analysis.**

```python
df.info()
```

Sentiment Bar consists 0 non-null (all are NaN), we can remove this column. Also, unicode codepoint won't be needed for the analysis it can be removed as well.

```python
emoji = df.drop(columns = ['Sentiment bar (c.i. 95%)', 'Unicode codepoint'])
emoji
```

**DC2: Rename the remaining columns using `snake_case` (all lowercase letters with underscores between words).**

```python
name_map = {
    'Char' : 'char',
    'Image [twemoji]' : 'image',
    'Occurrences [5...max]' : 'occurences',
    'Position [0...1]' : 'position',
    'Neg [0...1]' : 'negative',
    'Neut [0...1]' : 'neutral',
    'Pos [0...1]' : 'positive',
    'Unicode name' : 'unicode_name',
    'Unicode block' : 'unicode_block'
}

emoji = emoji.rename(columns = name_map)
emoji
```

**NV1: Add a new column called `sentiment`, where sentiment = (% positive tweets) - (% negative tweets).**

```python
emoji['sentiment'] = emoji.eval('positive - negative')
emoji
```

**NV2: Add a `positive_flag` column that is `True` if `sentiment > 0` (or above a set threshold), otherwise `False`.**

```python
emoji['positive_flag'] = emoji.eval('sentiment > 0')
emoji
```

**QUESTION 1: What percentage of emojis in the dataset have a positive sentiment?**

```python
emoji['positive_flag'].mean() * 100
```

**ANSWER** = 82.42%. (This code only count the "True" values. Since True = 1 and False = 0.)

**QUESTION 2: What percentage of the top 20 most popular emojis are positive?**

```python
top_20 = emoji.sort_values('occurences', ascending=False).head(20)
top_20['positive_flag'].sum() / len(top_20) * 100
```

**ANSWER** = 90% of the top 20 most popular emojis on tweets are flagged positive.

**QUESTION 3: Which emoji (with more than 500 mentions) is the most positive?**

```python
emo_500 = emoji.query('occurences > 500')
emo_500.sort_values('sentiment', ascending=False).head(1)
```

**ANSWER** = HEAVY BLACK HEART

**QUESTION 4: Which emoji (with more than 500 mentions) is the most negative?**

```python
emo_500 = emoji.query('occurences > 500')
emo_500.sort_values('sentiment', ascending=True).head(1)
```

**ANSWER** = UNAMUSED FACE

**QUESTION 5: Where in the tweets are most emojis located (i.e. at the beginning or the end)?**

```python
emoji['position'].mean()
```

**ANSWER** = Since a value near to 0 means that the position is at the beginning of a tweet(and 1 at the end). The resulting value 0.66555 (around 0.5) indicates that most emojis are located at the middle of the tweets.

**QUESTION 6: Is there a difference in the placement of positive versus negative emojis within a tweet?**

```python
#positive average
emoji.query('positive_flag == True')['position'].mean()
```

```python
#negative average
emoji.query('positive_flag == False')['position'].mean()
```

**ANSWER** = There is an approximately 0.02 difference in the position of positive versus negative emojis within a tweet. The values also imply that regardless of the sentiment value of the emojis, they are mostly placed at the middle of a tweet.
